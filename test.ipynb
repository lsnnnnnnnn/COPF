```python
from tgb.linkproppred.dataset_pyg import PyGLinkPropPredDataset

name = "tgbl-wiki"   # 或 "tgbl-review"
dataset = PyGLinkPropPredDataset(name=name, root="datasets")
data = dataset.get_TemporalData()

print(type(data), data.src.shape, data.dst.shape, data.t.shape)
print("has msg:", hasattr(data, "msg"), "msg shape:", getattr(data, "msg", None).shape if hasattr(data, "msg") else None)

# 时间是否非递减（online必需）
print("t nondecreasing:", bool((data.t[1:] >= data.t[:-1]).all()))

# 是否有 node_type（如果有，你就能直接用作 group）
print("has node_type:", dataset.node_type is not None)
if dataset.node_type is not None:
    nt = dataset.node_type
    print("node_type shape:", nt.shape, "unique:", sorted(set(nt.tolist()))[:10])

# split masks（你做 Pre/Deploy/Post 或 train/val/test）
print("train/val/test sizes:", int(dataset.train_mask.sum()), int(dataset.val_mask.sum()), int(dataset.test_mask.sum()))
```
